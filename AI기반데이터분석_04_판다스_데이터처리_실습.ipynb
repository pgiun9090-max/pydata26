## 1. 파이썬 파일 처리

```python
sample_file = 'data/sample.txt'
```

```python
# 파일을 open하여 전체 내용을 읽어서 출력
with open(sample_file, mode='rt', encoding='utf-8') as f:
    content = f.read()
    print(content)
print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기
with open(sample_file, mode='rt', encoding='utf-8') as f:
    for line in f:
        print(line, end='')
print("----- 끝 -----")
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옵니다. 따라서 print()로 출력하면 줄바꿈이 두 번 발생합니다.
with open(sample_file, mode='rt', encoding='utf-8') as f:
    for line in f:
        print(line, end='')
print("----- 끝 -----")
```

```python
# 여러 개의 파일 읽기 (모든 txt 파일을 읽어서 출력하기)
import glob

files = glob.glob("data/*.csv")
print(files)

for file in files:
    print(file)
    with open(file, mode="r") as f:
        print(f.read())
print("----- 끝 -----")
```

```python
score_df
```

```python
# 인덱서 이용 : 레이블(이름) 인덱서
# 행 접근
score_df.loc[1]
```

```python
# 열 접근
score_df.loc[:, ['이름', '국어']]
```

```python
# (행,열) 접근
score_df.loc[:1, '이름':'국어']
```

```python
# 인덱서 이용 : 위치(숫자) 인덱서
# 행 접근
score_df.iloc[:1]
```

```python
score_df
```

```python
# 조건 지정 추출 : Boolean Indexing
# 직접 지정 방식으로 행 추출

score_df[score_df.수학 >= 90]
```

```python
score_df[(score_df.국어 >= 90) & (score_df.수학 >= 90)]
```

```python
score_df[score_df.이름.isin(['정약용','이순신'])]
```

```python
# 인덱서 방식으로 행 추출
score_df.loc[score_df.수학 >= 90]
```

```python
# (행,열) 데이터 추출
score_df.loc[score_df.수학 >= 90, ['이름', '수학']]
```
